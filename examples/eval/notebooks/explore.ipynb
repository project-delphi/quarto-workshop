# A first look at the Bank Marketing data

Exploration before the evaluation was written: how many contacts, how many subscribed, and when.
The data are described in `../data/README.md`. Open this notebook from the project root with
`uv run --with jupyter jupyter lab notebooks/explore.ipynb`.

```python
import matplotlib.pyplot as plt
import pandas as pd

bank = pd.read_csv("../data/bank-full.csv", sep=";")
bank.shape
```

```python
bank.head()
```

The target `y` is `yes` when the client subscribed to a term deposit. The base rate is the share who did:

```python
subscribed = bank["y"] == "yes"
subscribed.mean()
```

Subscription rate by the month of the last contact, in calendar order, with the number of contacts:

```python
months = ["jan", "feb", "mar", "apr", "may", "jun",
          "jul", "aug", "sep", "oct", "nov", "dec"]
by_month = (subscribed.groupby(bank["month"]).agg(["mean", "size"])
            .reindex(months).rename(columns={"mean": "rate", "size": "contacts"}))
by_month
```

```python
fig, ax = plt.subplots(figsize=(7, 3))
ax.bar(by_month.index, by_month["rate"], color="#4C72B0")
ax.set_ylabel("Share who subscribed")
ax.set_title("Subscription rate by month of last contact")
plt.show()
```

Months with few contacts have high rates, so the month partly marks the campaign period.
`duration` (the call's length) is only known after the call, so the evaluation leaves it out.
